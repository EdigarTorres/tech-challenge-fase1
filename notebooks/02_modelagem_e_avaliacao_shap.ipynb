### 2. Justificativa Clínica da Métrica: Por que priorizar o Recall?
- **Falso Negativo (FN):** Classificar um tumor **Maligno** como **Benigno** é o erro mais grave na medicina diagnóstica, pois retarda o início do tratamento oncológico.
- **Escolha Técnica:** Priorizamos o **Recall (Sensibilidade)** da classe positiva (`Maligno = 1`) combinado ao **F1-Score** e **ROC-AUC** para manter alta precisão sem gerar alarmes falsos excessivos.

### 4. Discussão Crítica: O Modelo Pode Ser Utilizado na Prática Clínica?
**Sim, como ferramenta de triagem e apoio à decisão (CDSS - Clinical Decision Support System), nunca em substituição ao médico:**
1. **Priorização de Fila (Triagem Inteligente):** Exames com probabilidade elevada de malignidade ganham prioridade máxima na fila de revisão do médico patologista/oncologista.
2. **Segunda Opinião Explicável:** Ao exibir o gráfico **SHAP Waterfall** ao lado do laudo preliminar, o médico visualiza imediatamente quais variáveis morfológicas (`worst radius`, `worst concave points`, `worst perimeter`) motivaram o alerta.
3. **Soberania Médica:** O médico correlaciona o escore preditivo com anamnese, exame físico, histórico familiar e biópsia histopatológica antes de definir o diagnóstico definitivo.
